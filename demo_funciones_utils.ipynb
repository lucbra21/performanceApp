### 6.3 calcular_metricas(df, columnas, estadistica)
Función auxiliar para calcular estadísticas específicas sobre columnas dadas.

## 7. Funciones de Análisis Comparativo con Datos Históricos

### 7.1 get_historical_md_data(match_day_target, exclude_date=None)
Obtiene datos acumulados para un Match Day específico.

```python
# Obtener datos históricos para un Match Day específico
if df_gps is not None and 'Match Day' in df_gps.columns:
    # Obtener Match Days disponibles
    match_days_disponibles = df_gps['Match Day'].unique().to_list()
    print(f"Match Days disponibles: {match_days_disponibles}")
    
    # Usar el primer Match Day disponible como ejemplo
    if match_days_disponibles:
        match_day_ejemplo = match_days_disponibles[3]
        print(f"\nObteniendo datos históricos para: {match_day_ejemplo}")
        
        df_historico = get_historical_md_data(match_day_ejemplo)
        
        if df_historico is not None:
            print(f"Datos históricos obtenidos: {df_historico.height} registros")
            
            # Mostrar fechas únicas en los datos históricos
            if 'Date' in df_historico.columns:
                fechas_historicas = df_historico['Date'].unique().to_list()
                print(f"Fechas en datos históricos: {len(fechas_historicas)}")
                print(f"Primeras 3 fechas: {fechas_historicas[:3]}")
        else:
            print("No se encontraron datos históricos")
    else:
        print("No hay Match Days disponibles")
else:
    print("No hay datos GPS o columna Match Day para obtener datos históricos")
```

### 7.2 calcular_comparacion_fecha_md(df_fecha, estadistica_seleccionada, columnas_interes=None)
Calcula diferencias porcentuales y z-scores comparando datos de una fecha específica con el historial del mismo Match Day.

```python
# Demostrar análisis comparativo con datos históricos
if fecha_picker and df_fecha is not None:
    print(f"=== ANÁLISIS COMPARATIVO PARA FECHA: {fecha_picker} ===")
    
    # Usar solo algunas columnas para acelerar el ejemplo
    columnas_ejemplo = ['Distance (m)', 'Max Speed (km/h)']
    
    resultados_comparacion = calcular_comparacion_fecha_md(
        df_fecha, 
        estadistica_seleccionada='mean',
        columnas_interes=columnas_ejemplo
    )
    
    if resultados_comparacion:
        print(f"Análisis completado para {len(resultados_comparacion)} entidades")
        
        # Mostrar resultados para la primera entidad
        primera_entidad = list(resultados_comparacion.keys())[0]
        resultado_ejemplo = resultados_comparacion[primera_entidad]
        
        print(f"\nEjemplo de resultados para: {primera_entidad}")
        print(f"Tipo: {resultado_ejemplo.get('Player', resultado_ejemplo.get('Team', resultado_ejemplo.get('Position', 'Desconocido')))}")
        print(f"Match Day: {resultado_ejemplo.get('Match_Day')}")
        print(f"Fecha: {resultado_ejemplo.get('Date')}")
        
        print("\nDiferencias porcentuales:")
        for metrica, valor in resultado_ejemplo.get('diferencias_porcentuales', {}).items():
            print(f"  {metrica}: {valor}%" if valor is not None else f"  {metrica}: N/A")
        
        print("\nZ-scores:")
        for metrica, valor in resultado_ejemplo.get('z_scores', {}).items():
            print(f"  {metrica}: {valor}" if valor is not None else f"  {metrica}: N/A")
    else:
        print("No se pudieron calcular comparaciones")
else:
    print("No hay datos de fecha disponibles para análisis comparativo")
```
